```python
# 0 is for no diabetes or only during pregnancy, 1 is for pre-diabetes or borderline diabetes, 2 is for yes diabetes
df_selected['DIABETE3'] = df_selected['DIABETE3'].replace({2:0, 3:0, 1:2, 4:1})
df_selected = df_selected[df_selected.DIABETE3 != 7]
df_selected = df_selected[df_selected.DIABETE3 != 9]
df_selected.DIABETE3.unique()
```

```python
def _replace_and_filter(df, col, replace_map=None, drop_values=None):
    """
    Helper function to replace values and filter out unwanted rows.

    Args:
        df (pd.DataFrame): Input dataframe.
        col (str): Column name to process.
        replace_map (dict, optional): Mapping for value replacement.
        drop_values (list, optional): Values to remove from the column.

    Returns:
        pd.DataFrame: Updated dataframe.
    """
    if replace_map:
        df[col] = df[col].replace(replace_map)
    if drop_values:
        df = df[~df[col].isin(drop_values)]
    return df


def preprocess_health_data(df):
    """
    Cleans and preprocesses health survey data for downstream ML tasks.

    Transformations include:
    - Binary encoding of categorical variables
    - Removal of invalid/missing responses (e.g., 7, 9, 77, 99)
    - Normalization of BMI values
    - Preservation of ordinal features where appropriate

    Args:
        df (pd.DataFrame): Raw input dataframe.

    Returns:
        pd.DataFrame: Cleaned and processed dataframe.
    """

    df = df.copy()

    # Binary / categorical cleaning
    df = _replace_and_filter(df, '_RFHYPE5', {1: 0, 2: 1}, [9])
    df = _replace_and_filter(df, 'TOLDHI2', {2: 0}, [7, 9])
    df = _replace_and_filter(df, '_CHOLCHK', {2: 0, 3: 0}, [9])
    df = _replace_and_filter(df, 'SMOKE100', {2: 0}, [7, 9])
    df = _replace_and_filter(df, 'CVDSTRK3', {2: 0}, [7, 9])
    df = _replace_and_filter(df, '_MICHD', {2: 0})
    df = _replace_and_filter(df, '_TOTINDA', {2: 0}, [9])
    df = _replace_and_filter(df, '_FRTLT1', {2: 0}, [9])
    df = _replace_and_filter(df, '_VEGLT1', {2: 0}, [9])
    df = _replace_and_filter(df, '_RFDRHV5', {1: 0, 2: 1}, [9])
    df = _replace_and_filter(df, 'HLTHPLN1', {2: 0}, [7, 9])
    df = _replace_and_filter(df, 'MEDCOST', {2: 0}, [7, 9])
    df = _replace_and_filter(df, 'DIFFWALK', {2: 0}, [7, 9])
    df = _replace_and_filter(df, 'SEX', {2: 0})

    # Ordinal / filtered only
    df = _replace_and_filter(df, 'GENHLTH', drop_values=[7, 9])
    df = _replace_and_filter(df, '_AGEG5YR', drop_values=[14])

    # Mental & physical health days
    df = _replace_and_filter(df, 'MENTHLTH', {88: 0}, [77, 99])
    df = _replace_and_filter(df, 'PHYSHLTH', {88: 0}, [77, 99])

    # BMI scaling
    df['_BMI5'] = (df['_BMI5'] / 100).round(0)

    return df
```

```python
df_selected = preprocess_health_data(df_selected)
print(df_selected.shape)
df_selected.head()
```
